## Abstract

Maternal and women’s health remain historically under-represented fields of research, prompting us to ask how socioeconomic and demographic factors influence adverse pregnancy outcomes across the United States. Our central research question aimed to assess how socioeconomic and demographic factors influence adverse pregnancy outcomes nationally, and whether these national trends are reflected in maternal mortality patterns within a dense urban setting, such as New York City. We hypothesized that indicators of lower socioeconomic status, such as higher poverty rates and lower education, would be associated with increased adverse outcomes both nationally and locally. To investigate this, we conducted an analysis using two complementary sources: the comprehensive 2022-2023 National Survey of Family Growth (NSFG) dataset, which provided information on national adverse outcomes (e.g., miscarriage, abortion, stillbirth) alongside socioeconomic predictors (poverty, education, prenatal care status). For the localized case study, we performed an ecological analysis using NYC Pregnancy-Associated Mortality and Economic Conditions datasets to compare borough-level mortality and poverty rates. The analytical methodology involved statistical tests, like Chi-squared tests and ANOVA, culminating in a binary classifier model trained on the NSFG data to predict outcome classification.






The results demonstrated compelling and highly statistically significant relationships (p < 0.05) between adverse pregnancy outcomes and key socioeconomic factors across the national sample. It confirmed the predictive power of poverty level, educational attainment, labor force status, metropolitan area residence, and access to prenatal care. The model reinforced these statistical findings, revealing that lack of prenatal care engagement (got_prenatal_Unknown) was highly associated with an increased Abortion (coefficient 4.24), while generally, higher binned poverty levels correlated positively with the odds of both abortion and miscarriage. Mirroring these trends locally, the NYC dataset analysis confirmed stark geographic disparities in maternal mortality, showing that total pregnancy-related deaths were consistently highest in the Bronx and Brooklyn, compared to Manhattan and Staten Island. Additionally, there is a statistically significant difference in the mortality counts across different Boroughs. Furthermore, examining the most recent data (2021) demonstrated a visible trend where boroughs with higher poverty percentages also exhibited higher maternal death counts. This collective evidence supports the hypothesis, underscoring that economic stability and access to healthcare services are critical determinants of maternal and pregnancy health outcomes across diverse geographical scales.

### Dataset 3: Economic Data From New York

Dataset Information:

A.

A.
Description of Variables most relevant to the project


This dataset has information regarding the poverty rates in New York City, collected by the government of New York City, in the Environment and Health sector. This dataset has over 3000 observations and 7 variables, and has data from 2007 onwards to 2021. The key variables in this dataset include the Year or TimePeriod from when the data was collected, which is helpful for us, since we use the relevant time to join the poverty dates with our pregnancy mortality data. Next, the GeoType, GeoRank and GeoID columns carry information about the spatial locations from where the data was collected. We use this information to map our poverty rates from all over New York to the granularity that we have—the Boroughs. We use another dataset from the NYC government data to map NTA regions to the respective Boroughs. The last two variables are Number and Poverty. Number refers to the estimated number of people whose income falls below the level determined by the government, and Poverty estimates this number as a percent compared to the total number of people’s poverty status that is known. We will focus our analysis on the percent, since it is a better indicator of the comparative poverty level as opposed to an absolute number.


B. While this dataset is reliable and publicly maintained, it does have several limitations. First, the data is reported across many different geographic units (e.g., NTA2020, UHF zones, community districts), and these boundaries are not always consistent with one another. Because our analysis requires borough-level poverty estimates, we must drop older or incompatible GeoTypes and perform additional mapping, which introduces the possibility of geographic mismatches. Second, the dataset only provides poverty counts and percentages without information on margins of error or population sample sizes, meaning we cannot assess the precision of estimates across neighborhoods. Finally, the dataset reflects conditions only in New York City, which limits its generalizability beyond the city and requires us to be cautious when comparing these values to maternal health outcomes that may vary across broader regions.

```python
# Keep only geographic units we can reliably map to boroughs
map_df = df[df['GeoType'].isin(['nta2020', 'borough', 'citywide'])].copy()
```

```python
#clean column

def clean_name(x):
    if pd.isna(x):
        return x
    x = x.replace("–", "-").replace("—", "-")
    x = x.strip().title()
    return x


map_df['GeoType'] = map_df['GeoType'].str.lower().str.strip()
map_df['Geography_clean'] = map_df['Geography'].apply(clean_name)
```

```python
#boroughs
borough_df = map_df[map_df['GeoType'] == 'borough'].copy()
borough_df['BoroName_clean'] = borough_df['Geography_clean']
```

```python
#citywide
citywide_df = map_df[map_df['GeoType'] == 'citywide'].copy()
citywide_df['BoroName_clean'] = "New York City"
```

```python
#combining all
df_clean = pd.concat([nta_df, borough_df, citywide_df], ignore_index=True)

#final borough column
df_clean['borough'] = (
    df_clean['BoroName_clean']
      .str.lower()
      .str.replace(" ", "_")
      .str.replace("-", "_")
)


#dropping temp columns
df_clean = df_clean.drop(columns=[
    'Geography_clean',
    'NTAName_clean',
    'BoroName_clean'
])
```

At this point, a learning curve can tell us a lot about our model and whether we are underfitting or overfitting, and if more data could help our analysis.

```python
from sklearn.model_selection import learning_curve

def plot_learning_curve(estimator, title, X, y, ylim=None, cv=5,
                        train_sizes=np.linspace(.1, 1.0, 5)):
    plt.figure(figsize=(10, 6))
    plt.title(title)
    if ylim is not None:
        plt.ylim(*ylim)
    plt.xlabel("Training examples")
    plt.ylabel("Recall Score")

    train_sizes, train_scores, test_scores = learning_curve(
        estimator,
        X,
        y,
        cv=cv,
        scoring='recall',
        n_jobs=1
    )

    train_scores_mean = np.mean(train_scores, axis=1)
    train_scores_std = np.std(train_scores, axis=1)
    test_scores_mean = np.mean(test_scores, axis=1)
    test_scores_std = np.std(test_scores, axis=1)

    plt.grid()

    # Plot the shaded confidence areas
    plt.fill_between(train_sizes, train_scores_mean - train_scores_std,
                     train_scores_mean + train_scores_std, alpha=0.1, color="r")
    plt.fill_between(train_sizes, test_scores_mean - test_scores_std,
                     test_scores_mean + test_scores_std, alpha=0.1, color="g")

    # Plot the lines
    plt.plot(train_sizes, train_scores_mean, 'o-', color="r",
             label="Training score")
    plt.plot(train_sizes, test_scores_mean, 'o-', color="g",
             label="Cross-validation score")

    plt.legend(loc="best")
    return plt

plot_learning_curve(
    binary_model,
    "Learning Curve (Random Forest)",
    X_train.values,
    y_train_binary.values
)
plt.show()
```

**Analysis of Learning Curve:**
The learning curve demonstrates that our Random Forest model reaches a stability plateau with a Recall score of approximately **0.625 (62.5%)** on the validation set.

1.  **Diminishing Returns:** Both the training and cross-validation scores flatten out as the training set size increases. This indicates that sample size is no longer the limiting factor. Simply collecting more rows of similar demographic data would likely not improve the model's performance.
2.  **Feature Limitations:** The convergence of these scores suggests that the remaining error is due to irreducible error or missing predictors. While socioeconomic factors (income, education) and basic health history are predictive, they cannot account for the remaining ~37.5% of variance in adverse outcomes. To improve the model further, we would need to incorporate granular clinical data (e.g., genetic screenings, daily vitals) rather than just survey responses.
3.  **Generalization:** The gap between the training score (0.73) and validation score (0.625) is moderate, suggesting the model is slightly overfitted but generally robust.

This may imply that access to specific types of care (captured by `got_prenatal`) is a more critical determinant than insurance coverage itself.

**Predictive Modeling: Random Forest Classifier**
To capture complex, non-linear interactions between these variables, we transitioned from a baseline logistic regression to a **Random Forest Classifier**. The model was trained to predict a binary target: *Live Birth* (0) vs. *Adverse Outcome* (1) (e.g., miscarriage, stillbirth).

The Random Forest model achieved an overall **accuracy of 58%** on the test set. Breaking this down by class:
* **Live Births:** The model performed well, with a precision of **0.80**, indicating a high success rate when predicting positive outcomes.
* **Adverse Outcomes:** The model achieved a **recall of 0.66** and a **precision of 0.37**.

**Interpretation of Model Performance**
The recall of 0.66 for adverse outcomes indicates that our model successfully identifies more than half of the at-risk cases using only socioeconomic and demographic data. However, the lower precision (0.37) implies a high false-positive rate; the model often flags pregnancies as "at risk" that result in live births.

In a medical screening context, this trade-off is often acceptable, as identifying potential risks (high recall) is prioritized over avoiding false alarms. The results validate that while socioeconomic factors like poverty and education are strong signals, they do not tell the whole story. Clinical and biological factors likely account for the remaining variance in pregnancy outcomes.

**Feature Importance Analysis**
To understand which factors drove the Random Forest model’s predictions, we analyzed the Gini importance scores for each feature. The results highlight a distinct hierarchy of influence:

1.  **Primary Drivers:**
    **Education** (0.217) and **Poverty Status** (0.208) were some overwhelmingly dominant features. This indicates that a mother's socioeconomic background is a strong signal for predicting pregnancy outcomes in our model, far outweighing insurance or geography.
    **Prenatal Care Access** (`got_prenatal`) was the most critical feature (0.257). Receiving proper prenatal care seems to be extremely important for pregnancy outcome. All together, these features jointly account for the majority of the total feature importance.
2.  **Low-Impact Features:**
    Consistent with our earlier statistical tests, **Insurance Status** and **Metropolitan Area** played minor roles (importance scores < 0.05). This reinforces the finding that the *type* of insurance coverage is less predictive of birth outcomes than the underlying economic stability (poverty/education) of the mother.

**Conclusion:**
The model effectively prioritized long-term structural determinants (Education, Poverty, Healthcare Access) over short-term transactional variables (Insurance Type), validating the hypothesis that significant long-term socioeconomic status could be the primary predictors of pregnancy outcomes in this dataset.
